```python
# ==========================================================
# EXPLANATORY ANALYSIS NOTEBOOK
# EduProfile AI — Business Understanding & Explanatory Analysis
# ==========================================================
# File Name : explanatory.ipynb
# Tujuan :
# Notebook ini digunakan untuk menjawab business question
# menggunakan explanatory analysis berbasis data.
#
# Jalankan setelah:
# python capstone.py
#
# Dataset:
# output/final_dataset_cleaned_with_pace.csv
# ==========================================================

# ==========================================================
# 1. IMPORT LIBRARIES
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', font_scale=1.1)

# ==========================================================
# 2. LOAD DATASET
# ==========================================================

df = pd.read_csv('output/final_dataset_cleaned_with_pace.csv')

print("Dataset Loaded Successfully")
print("Shape :", df.shape)

display(df.head())

# ==========================================================
# 3. BUSINESS UNDERSTANDING
# ==========================================================

print("""
BUSINESS BACKGROUND
-------------------------------------------------
EduProfile AI merupakan sistem analitik pendidikan
yang bertujuan membantu personalisasi pembelajaran
berdasarkan gaya belajar siswa (Visual, Auditory,
Kinesthetic) dan learning pace siswa.

Dengan memahami pola belajar siswa, institusi
pendidikan dapat memberikan strategi belajar yang
lebih efektif dan adaptif.
""")

# ==========================================================
# 4. BUSINESS QUESTIONS
# ==========================================================

business_questions = [
    "1. Bagaimana distribusi learning style siswa?",
    "2. Faktor apa yang paling memengaruhi learning style?",
    "3. Bagaimana hubungan learning pace dengan performa akademik?",
    "4. Apakah indikator VAK mampu membedakan gaya belajar siswa?",
    "5. Bagaimana pola hubungan antar fitur numerik?",
]

print("BUSINESS QUESTIONS")
print("-------------------------------------------------")

for q in business_questions:
    print(q)

# ==========================================================
# 5. DATA UNDERSTANDING
# ==========================================================

print("\nDATA UNDERSTANDING")
print("-------------------------------------------------")

print("Jumlah Baris :", df.shape[0])
print("Jumlah Kolom :", df.shape[1])

print("\nNama Kolom:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

# ==========================================================
# 6. LABEL LEARNING PACE
# ==========================================================

cluster_means = (
    df.groupby('LearningPace_Cluster')['AcademicScore']
    .mean()
    .sort_values()
)

cluster_map = {
    cluster_means.index[0]: 'Slow',
    cluster_means.index[1]: 'Medium',
    cluster_means.index[2]: 'Fast'
}

df['LearningPace_Label'] = (
    df['LearningPace_Cluster']
    .map(cluster_map)
)

print("\nLearning Pace Label Mapping:")
print(cluster_map)

# ==========================================================
# 7. DESCRIPTIVE STATISTICS
# ==========================================================

print("\nDESCRIPTIVE STATISTICS")
print("-------------------------------------------------")

display(df.describe())

# ==========================================================
# BUSINESS QUESTION 1
# Bagaimana distribusi learning style siswa?
# ==========================================================

print("""
==========================================================
BUSINESS QUESTION 1
Bagaimana distribusi learning style siswa?
==========================================================
""")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Pie chart
style_counts = df['LearningStyle'].value_counts()

axes[0].pie(
    style_counts,
    labels=style_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

axes[0].set_title('Distribusi Learning Style')

# Countplot
sns.countplot(
    data=df,
    x='LearningStyle',
    ax=axes[1]
)

axes[1].set_title('Jumlah Siswa per Learning Style')

plt.tight_layout()
plt.show()

print("\nINSIGHT:")
print("- Distribusi learning style relatif seimbang.")
print("- Dataset ideal digunakan untuk modeling klasifikasi.")
print("- Tidak terdapat dominasi ekstrem pada salah satu kelas.")

# ==========================================================
# BUSINESS QUESTION 2
# Faktor apa yang paling memengaruhi learning style?
# ==========================================================

print("""
==========================================================
BUSINESS QUESTION 2
Faktor apa yang paling memengaruhi learning style?
==========================================================
""")

feature_cols = [
    'AcademicScore',
    'AttendanceRate',
    'StudentPerformance',
    'DeviceUsage',
    'Indikator_Visual',
    'Indikator_Auditory',
    'Indikator_Kinestetik'
]

group_mean = (
    df.groupby('LearningStyle')[feature_cols]
    .mean()
    .round(2)
)

display(group_mean)

# Heatmap rata-rata
plt.figure(figsize=(10, 6))

sns.heatmap(
    group_mean,
    annot=True,
    cmap='coolwarm'
)

plt.title('Rata-rata Fitur per Learning Style')
plt.show()

print("\nINSIGHT:")
print("- Setiap learning style memiliki karakteristik berbeda.")
print("- Indikator Visual tinggi pada siswa Visual.")
print("- Indikator Auditory tinggi pada siswa Auditory.")
print("- Indikator Kinestetik tinggi pada siswa Kinesthetic.")

# ==========================================================
# BUSINESS QUESTION 3
# Bagaimana hubungan learning pace dengan performa akademik?
# ==========================================================

print("""
==========================================================
BUSINESS QUESTION 3
Bagaimana hubungan learning pace dengan performa akademik?
==========================================================
""")

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='LearningPace_Label',
    y='AcademicScore',
    order=['Slow', 'Medium', 'Fast']
)

plt.title('Academic Score per Learning Pace')
plt.show()

pace_stats = (
    df.groupby('LearningPace_Label')['AcademicScore']
    .agg(['mean', 'std'])
    .round(2)
)

display(pace_stats)

print("\nINSIGHT:")
print("- Cluster Fast memiliki AcademicScore tertinggi.")
print("- Cluster Slow memiliki performa akademik lebih rendah.")
print("- Learning pace dapat digunakan untuk personalisasi pembelajaran.")

# ==========================================================
# BUSINESS QUESTION 4
# Apakah indikator VAK mampu membedakan gaya belajar siswa?
# ==========================================================

print("""
==========================================================
BUSINESS QUESTION 4
Apakah indikator VAK mampu membedakan gaya belajar siswa?
==========================================================
""")

indikators = [
    'Indikator_Visual',
    'Indikator_Auditory',
    'Indikator_Kinestetik'
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, col in enumerate(indikators):

    sns.violinplot(
        data=df,
        x='LearningStyle',
        y=col,
        ax=axes[i]
    )

    axes[i].set_title(col)

plt.tight_layout()
plt.show()

print("\nINSIGHT:")
print("- Distribusi indikator berbeda pada tiap learning style.")
print("- Indikator VAK mampu memisahkan pola belajar siswa.")
print("- Feature engineering berhasil membentuk representasi gaya belajar.")

# ==========================================================
# BUSINESS QUESTION 5
# Bagaimana hubungan antar fitur numerik?
# ==========================================================

print("""
==========================================================
BUSINESS QUESTION 5
Bagaimana hubungan antar fitur numerik?
==========================================================
""")

corr = df[feature_cols].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()

print("\nINSIGHT:")
print("- Tidak terdapat multikolinearitas ekstrem.")
print("- Sebagian besar fitur memiliki korelasi moderat.")
print("- Feature aman digunakan untuk modeling machine learning.")

# ==========================================================
# 8. PCA VISUALIZATION
# ==========================================================

print("""
==========================================================
PCA VISUALIZATION
==========================================================
""")

pace_features = [
    'AcademicScore',
    'AttendanceRate',
    'DeviceUsage'
]

X = df[pace_features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)

components = pca.fit_transform(X_scaled)

df['PCA1'] = components[:, 0]
df['PCA2'] = components[:, 1]

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='PCA1',
    y='PCA2',
    hue='LearningPace_Label',
    alpha=0.6
)

plt.title('PCA Visualization of Learning Pace')
plt.show()

print("\nINSIGHT:")
print("- Cluster learning pace terlihat cukup terpisah.")
print("- K-Means berhasil membentuk segmentasi learning pace.")
print("- PCA membantu visualisasi pola persebaran siswa.")

# ==========================================================
# 9. FINAL CONCLUSION
# ==========================================================

print("""
==========================================================
FINAL CONCLUSION
==========================================================
""")

print("""
1. Distribusi learning style siswa relatif seimbang,
   sehingga dataset layak digunakan untuk modeling AI.

2. Feature engineering indikator VAK berhasil
   merepresentasikan karakteristik tiap gaya belajar.

3. Learning pace memiliki hubungan terhadap performa akademik,
   di mana cluster Fast memiliki nilai akademik lebih tinggi.

4. Korelasi antar fitur tergolong aman dan tidak menunjukkan
   multikolinearitas ekstrem.

5. Sistem EduProfile AI berpotensi membantu personalisasi
   pembelajaran berbasis data dan AI.
""")

# ==========================================================
# 10. RECOMMENDATION
# ==========================================================

print("""
==========================================================
BUSINESS RECOMMENDATION
==========================================================
""")

print("""
1. Sekolah dapat menerapkan personalisasi materi belajar
   berdasarkan learning style siswa.

2. Siswa dengan learning pace Slow dapat diberikan
   adaptive learning dan pendampingan tambahan.

3. Institusi pendidikan dapat menggunakan model AI
   untuk mendeteksi gaya belajar siswa secara otomatis.

4. Dashboard monitoring dapat digunakan guru untuk
   memantau perkembangan siswa secara real-time.
""")

print("\nEXPLANATORY ANALYSIS COMPLETED SUCCESSFULLY")
```
